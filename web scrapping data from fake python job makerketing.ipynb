# Job Market Intelligence Dashboard Using Web Scraping

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://realpython.github.io/fake-jobs/"

html = requests.get(url).text

soap = BeautifulSoup(html, "html.parser")
soap
```

```python
# 1. Find all the job cards on the page
job_elements = soap.find_all("div", class_="card-content")

jobs = []

# 2. Loop through each card and extract the details
for job_element in job_elements:
    title_element = job_element.find("h2", class_="title")
    company_element = job_element.find("h3", class_="company")
    location_element = job_element.find("p", class_="location")
    time_element = job_element.find("time")
    
    # Extract the links (the second 'a' tag usually points to the Apply link)
    links = job_element.find_all("a")
    apply_link = links[1]["href"] if len(links) > 1 else None
    
    # Clean the text data and append to your list
    jobs.append([
        title_element.text.strip(),
        company_element.text.strip(),
        location_element.text.strip(),
        time_element.text.strip(),
        "Fake Job Description",  # The main page doesn't show full descriptions
        apply_link
    ])

# 3. Create your DataFrame
df = pd.DataFrame(
    jobs,
    columns=["job_title", "company", "location", "posting date", "description", "apply link"]
)

df
```

## Data Cleaning

```python
df.dtypes
```

##### Convert data types

```python
df['job_title'] = df['job_title'].astype(str)
df['company'] = df['company'].astype(str)
df['location'] = df['location'].astype(str)
df['posting date'] = pd.to_datetime(df['posting date'])
df['description'] = df['description'].astype(str)
```

## Analysing
